```python
import gymnasium as gym


# ---------------------------------------------------------
# RANDOM AGENT FUNCTION
# ---------------------------------------------------------

def random_agent(environment_name):

    print("\n" + "=" * 60)
    print("ENVIRONMENT:", environment_name)
    print("=" * 60)

    # 1. Create the environment
    env = gym.make(environment_name)

    # 2. Reset the environment
    state, info = env.reset()

    # Variables
    total_reward = 0
    steps = 0

    # 3. Observe initial state
    print("\nInitial State:", state)
    print("Action Space:", env.action_space)

    # 4-8. Select random action and interact with environment
    while True:

        # 5. Select an action randomly
        action = env.action_space.sample()

        # 6. Execute the action
        next_state, reward, terminated, truncated, info = env.step(action)

        # 7. Observe next state and reward
        steps += 1
        total_reward += reward

        print("\nStep:", steps)
        print("Current State:", state)
        print("Random Action:", action)
        print("Reward:", reward)
        print("Next State:", next_state)

        # Update state
        state = next_state

        # 8. Check whether episode has terminated
        if terminated or truncated:
            break

    # 9. Calculate average reward
    average_reward = total_reward / steps

    # Display results
    print("\n" + "-" * 60)
    print("EPISODE COMPLETED")
    print("-" * 60)
    print("Environment       :", environment_name)
    print("Total Steps       :", steps)
    print("Total Reward      :", total_reward)
    print("Average Reward    :", round(average_reward, 4))

    # Close environment
    env.close()


# ---------------------------------------------------------
# RUN RANDOM AGENT FOR CARTPOLE
# ---------------------------------------------------------

random_agent("CartPole-v1")


# ---------------------------------------------------------
# RUN RANDOM AGENT FOR MOUNTAINCAR
# ---------------------------------------------------------

random_agent("MountainCar-v0")


# ---------------------------------------------------------
# RUN RANDOM AGENT FOR FROZENLAKE
# ---------------------------------------------------------

random_agent("FrozenLake-v1")
```
