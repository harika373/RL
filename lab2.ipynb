```python
import random

machines = ['A', 'B', 'C']

N_play = 10

epsilon = 0.2

N = {
    'A': 0,
    'B': 0,
    'C': 0
}

R = {
    'A': 0,
    'B': 0,
    'C': 0
}

Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

total_reward = 0

for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:
        selected_machine = random.choice(machines)
        decision = "Exploration"

    else:
        max_q = max(Q.values())

        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    print("\n--------------------------------")
    print(f"Play {play}")
    print(f"Random number = {r:.4f}")
    print(f"Decision = {decision}")
    print(f"Selected Machine = {selected_machine}")

    reward = float(
        input(f"Enter reward obtained from Machine {selected_machine}: ")
    )

    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine])
        / N[selected_machine]
    )
    total_reward += reward

    print(f"Reward received = {reward}")
    print(f"Updated Q({selected_machine}) = "
          f"{Q[selected_machine]:.2f}")


print("\n\n========================================")
print("FINAL RESULTS")
print("========================================")

print("\nN Table - Participation Count")
print("--------------------------------")
for machine in machines:
    print(f"Machine {machine}: {N[machine]} plays")

print("\nR Table - Total Rewards")
print("--------------------------------")
for machine in machines:
    print(f"Machine {machine}: {R[machine]:.2f}")

print("\nQ Table - Estimated Values")
print("--------------------------------")
for machine in machines:
    print(f"Q({machine}) = {Q[machine]:.2f}")

print("\n========================================")
print(f"Total Reward after {N_play} plays = {total_reward:.2f}")
print("========================================")
```
