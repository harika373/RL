```python
n = int(input("Enter the number of machines: "))

machines = []
rewards = {}


for i in range(n):
    machine = input(f"\nEnter name of Machine {i + 1}: ")
    machines.append(machine)

    reward_input = input(f"Enter rewards for Machine {machine} separated by spaces: ")
    rewards[machine] = list(map(float, reward_input.split()))


q_values = {}

for machine in machines:
    q_values[machine] = sum(rewards[machine]) / len(rewards[machine])


print("\n--- Estimated Q Values ---")

for machine in machines:
    print(f"Machine {machine}")
    print(f"Rewards: {rewards[machine]}")
    print(f"Q({machine}) = {q_values[machine]:.2f}")

selected_machine = max(q_values, key=q_values.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```

```python
n = int(input("Enter the number of machines: "))

machines = []
rewards = {}

for i in range(n):
    machine = input(f"\nEnter name of Machine {i + 1}: ")

    reward_input = input(
        f"Enter rewards for Machine {machine} separated by spaces: "
    )

    machines.append(machine)
    rewards[machine] = list(map(float, reward_input.split()))


Q = {}
N = {}

for machine in machines:
    Q[machine] = 0
    N[machine] = 0

print("\n--- Incremental Updates ---")


for machine in machines:
    for reward in rewards[machine]:


        N[machine] += 1


        Q[machine] = Q[machine] + (
            reward - Q[machine]
        ) / N[machine]

        print(
            f"Machine {machine}, "
            f"Reward = {reward}, "
            f"N = {N[machine]}, "
            f"Q({machine}) = {Q[machine]:.2f}"
        )


print("\n--- Final Q Values ---")

for machine in machines:
    print(f"Machine {machine}")
    print(f"N({machine}) = {N[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")


selected_machine = max(Q, key=Q.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```
